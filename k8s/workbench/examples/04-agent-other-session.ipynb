# 04 · Another scientist's agent session (bio scientist)

A supervisor answers only **their own** sessions. This notebook tries to read and approve a
session the control scientist started (notebook 03, step 2) and shows the refusals.
Exercise E48 in the platform manual follows it.

You need the control scientist's session id and gate id.

```python
from medcea_agent import Client, CliError

c = Client()
c.signin()          # open the link, sign in, confirm the code; the cell waits
```

```python
sid2 = "<the control scientist's session id>"
gid = "<the gate id>"
```

Each of these must be refused with **404 "no such session"**: the orchestrator answers
as if the session did not exist, so nobody learns another person's session ids or gates.

```python
for what, call in [
    ("read its state", lambda: c.info(sid2)),
    ("read its events", lambda: c.events(sid2)),
    ("approve its gate", lambda: c.answer(sid2, "gates", gid, "approve", note="E48: must be refused")),
    ("stop it", lambda: c.stop(sid2)),
]:
    try:
        call()
        print("NOT REFUSED:", what)
    except CliError as e:
        print("refused   :", what, "->", e)
```
